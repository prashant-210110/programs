```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df=sns.load_dataset('tips')
df.head()
sns.scatterplot(x='total_bill',y='tip',data=df)
plt.show()
```

```python
sns.histplot(df['total_bill'],kde=True,color='red')
```

```python
data={
    "class":["a","b","c","d","e"],
    "score":[10,20,30,40,50]
}
df=pd.DataFrame(data)
sns.boxplot(x='class',y='score',data=df)
```

```python
data={
    "score":[10,20,30,40,50]
}
df=pd.DataFrame(data)
sns.heatmap(df.corr(),annot=True)
```

```python
df=sns.load_dataset('iris')
sns.pairplot(data=df)
```

```python
df=sns.load_dataset('tips')
sns.countplot(x="day",data=df)
```

```python
data = {
    "department": ["HR","HR","IT","IT","Finance","Finance"],
    "salary": [3000, 3500, 6000, 6500, 5000, 5200]
}
df = pd.DataFrame(data)

sns.violinplot(x="department", y="salary", data=df)


```

```python
df=sns.load_dataset('tips')
sns.set_style("darkgrid")
plt.scatter(x="total_bill",y="tip",data=df)
```

```python
df=sns.load_dataset('tips')
sns.regplot(x="total_bill", y="tip", data=df, scatter_kws={'alpha':0.5})
```
