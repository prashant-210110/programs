```python
from nltk import CountVectoriser
from sklearn.metrics.pairwise import cosine_similarity
def is_duplicated(question1, question2):
    vectorizer = CountVectorizer().fit_transform([question1, question2])
    vectors = vectorizer.toarray()
    cosine_sim = cosine_similarity(vectors)
    if cosine_sim[0][1] > 0.8:
        return "true"
    else:
        return "false"
question1=input("enter the first question")
question2=input("enter the second question")    
is_duplicated(question1, question2)
```
